```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import altair as alt
```

```python
jekyll_path = '/Users/slbha/Desktop/is445/sreelayabhamidi.github.io/assets/json/'
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/ufo-scrubbed-geocoded-time-standardized-00.csv', 
                 names = ['date_time', 'city', 'state', 'country', 'shape', 'duration_seconds', 'duration_uncoded', 
                            'description', 'reported_date', 'latitude', 'longitude'],
                            na_values = {
                                'date_time': None,
                                'city': None,
                                'shape': None,
                                'country': None,
                                'shape': None,
                                'duration_seconds': None,
                                'duration_uncoded': None,
                                'description': None,
                                'reported_date': None,
                                'latitude': None,
                                'longitude': None
                            }, parse_dates=['date_time'])
df
```

```python
df.isna().sum()
```

```python
alt.data_transformers.disable_max_rows()
```

```python
grouped_by_shape = df.groupby('shape')['duration_seconds'].mean().reset_index()
grouped_by_shape
```

```python
shape_bars = alt.Chart(grouped_by_shape).mark_bar().encode(x = alt.X('shape:N'), y = alt.Y('duration_seconds:Q', scale = alt.Scale(type = 'log')), 
                                                               color = alt.Color('duration_seconds:Q', scale = alt.Scale(scheme = 'bluepurple')))
shape_bars
```

```python
shape_bars.save(jekyll_path + 'shape_bars.json')
```

```python
country = df[['date_time', 'country', 'duration_seconds']]
country['country'].unique()
```

```python
dropdown = alt.binding_select(options = ['us', 'gb', 'ca', 'au', 'de'], name = 'Select Country: ')
selection = alt.selection_point(fields = ['country'], value = 'us', bind = dropdown)
country_chart = alt.Chart(country).mark_point(color = 'darkgreen').encode(x = alt.X('date_time:T'), y = alt.Y('duration_seconds:Q', scale = alt.Scale(type = 'log')), 
                                             opacity = alt.condition(selection, alt.value(1), alt.value(0))).add_params(selection)
country_chart
```

```python
country_chart.save(jekyll_path + 'country_dropdown.json')
```
